```python
import pandas as pd
import openpyxl
from openpyxl.styles import PatternFill

file_list  = ['2022_年間売上表.xlsx','2023_年間売上表.xlsx']

df =pd.DataFrame()

for file_name in file_list:
    temp_df = pd.read_excel(file_name)
    df = pd.concat([df,temp_df],ignore_index = True)

result = df.groupby(['商品','売上年'],as_index = False).sum()

writer = pd.ExcelWriter('売上集計表.xlsx')
result.to_excel(writer,sheet_name = '集計表',index = False)
writer.close()

wb = openpyxl.load_workbook('売上集計表.xlsx')
ws = wb['集計表']

for row in ws['A1:C1']:
    for cell in row:
        cell.fill =  PatternFill(fill_type = 'solid',fgColor = 'F2F2F2')

wb.save('売上集計表.xlsx')




```
