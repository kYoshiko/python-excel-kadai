```python
import openpyxl
import datetime

today = datetime.date.today()
workbook = openpyxl.Workbook()
ws = workbook.active

title = '請求書'
company_name = '株式会社ABC'
post_code = '101-0022'
company_address = '東京都千代田区神田練塀町300'
phone_number = '03-1234-5678'
fax_number = '03-1234-5678'
person_in_charge = '鈴木一郎'
number = 1
billing_date = datetime.datetime.strftime(today, '%Y/%m/%d')

ws['B2'] = title
ws['B4'] = company_name
ws['B5'] = f'〒{post_code} {company_address}'
ws['B6'] = f'TEL:{phone_number} FAX:{fax_number}'
ws['B7'] = f'担当者名:{person_in_charge} 様'

ws['F4'] = 'No.'
ws['G4'] = str(number).zfill(4)
ws['F5'] = '日付'
ws['G5'] = billing_date

header = ['商品名','数量','単価','金額']

header_row = ws.max_row+3
i = 2
for h in header:
    ws.cell(row = header_row,column=i).value = h
    i = i + 1

data = [
   ['商品A',2 , 10000 ] ,
   ['商品B',1 , 15000 ] ,
]

start_row = 11
start_col = 2

for r_idx, row_data in enumerate(data):
    for c_idx, val in enumerate(row_data):
        ws.cell(row = start_row + r_idx , column = start_col + c_idx , value = val)

total_amount = 0
for r_idx, row_data in enumerate(data):
    amount = ws.cell(row = start_row + r_idx , column = 3).value * ws.cell(row = start_row + r_idx , column = 4).value
    ws.cell(row = start_row + r_idx , column = 5 , value = amount)
    total_amount = total_amount + amount

last_row = ws.max_row
ws.cell(row = last_row + 1 , column = 5 , value = total_amount)
tax = total_amount * 0.1
total_price = total_amount + tax
ws.cell(row = last_row + 3 , column = 2 , value = '合計')
ws.cell(row = last_row + 3 , column = 5 , value = total_amount)
ws.cell(row = last_row + 4 , column = 2 , value = '消費税')
ws.cell(row = last_row + 4 , column = 5 , value = tax)
ws.cell(row = last_row + 5 , column = 2 , value = '税込合計')
ws.cell(row = last_row + 5 , column = 5 , value = total_price)

workbook.save(f'請求書_{datetime.datetime.strftime(today, '%Y%m%d')}.xlsx')
```
